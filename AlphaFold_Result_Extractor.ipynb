```python
from google.colab import files
import zipfile
import os
import glob
import json
import pandas as pd


# ===============================
# Upload AlphaFold zip
# ===============================

uploaded = files.upload()

zip_file = list(uploaded.keys())[0]


BASE_DIR = "/content/alphafold_results"


os.makedirs(
    BASE_DIR,
    exist_ok=True
)


with zipfile.ZipFile(
    zip_file,
    "r"
) as z:

    z.extractall(
        BASE_DIR
    )


print("Extracted")



# ===============================
# Find confidence files
# ===============================


json_files = glob.glob(
    BASE_DIR +
    "/**/*summary_confidences_0.json",
    recursive=True
)


cif_files = glob.glob(
    BASE_DIR +
    "/**/*model_0.cif",
    recursive=True
)



print(
    "Confidence JSON:",
    len(json_files)
)


print(
    "CIF models:",
    len(cif_files)
)



# ===============================
# Extract AlphaFold metrics
# ===============================


records = []


for js in json_files:


    folder = os.path.basename(
        os.path.dirname(js)
    )


    with open(
        js,
        "r"
    ) as f:

        data = json.load(f)



    records.append(

        {

        "generated_id":
            folder.replace(
                "pg2pet_",
                "PG2PET_"
            ).upper(),


        "folder":
            folder,


        "ptm":
            data.get(
                "ptm",
                None
            ),


        "ranking_score":
            data.get(
                "ranking_score",
                None
            ),


        "fraction_disordered":
            data.get(
                "fraction_disordered",
                None
            ),


        "has_clash":
            data.get(
                "has_clash",
                None
            ),


        "cif_file":
            next(
                (
                x for x in cif_files
                if folder in x
                ),
                None
            )

        }

    )



df = pd.DataFrame(
    records
)



display(df)



# ===============================
# Save
# ===============================


output = (
    "/content/AlphaFold_structure_summary.xlsx"
)


df.to_excel(
    output,
    index=False
)


print(
    "Saved:",
    output
)
```
